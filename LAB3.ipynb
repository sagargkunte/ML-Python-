```python
import numpy as np
import matplotlib.pyplot as plt
from sklearn import datasets
from sklearn.preprocessing import StandardScaler

# Load data
iris = datasets.load_iris()
X, y = iris.data, iris.target

# Standardize
X = StandardScaler().fit_transform(X)

# Covariance + Eigen
cov = np.cov(X.T)
eig_vals, eig_vecs = np.linalg.eig(cov)

print("Covariance Matrix:\n", cov)
print("\nEigen Values:\n", eig_vals)

# -------- 3D Plot --------
fig = plt.figure()
ax = fig.add_subplot(111, projection='3d')

colors = ['r', 'g', 'b']
for i in range(3):
    ax.scatter(X[y==i,0], X[y==i,1], X[y==i,2], c=colors[i])

ax.set_title("Iris Data Before PCA")
plt.show()

# -------- Eigen Vectors --------
fig = plt.figure()
ax = fig.add_subplot(111, projection='3d')

for i in range(3):
    ax.scatter(X[y==i,0], X[y==i,1], X[y==i,2], c=colors[i])

for i in range(3):
    ax.quiver(0,0,0,
              eig_vecs[0,i],
              eig_vecs[1,i],
              eig_vecs[2,i],
              color='k')

ax.set_title("Eigen Vectors on Iris Data")
plt.show()
```
