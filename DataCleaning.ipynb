```python
# !pip install scikit-learn

from sklearn.preprocessing import LabelEncoder
import pandas as pd
df = pd.read_csv("./Datasets/Student Depression Dataset.csv")
print(df)
df_label = df.copy()

le = LabelEncoder()
df_label['Gender_Encoded'] = le.fit_transform(df_label['Gender'])

print('\n Label encoded data')
print(df_label[['Gender_Encoded']].head())
# Till this label encoding (Label encoding can be done only for those who has yes/no values or True/False values etc.. which has only 2 values)
print("-----------------------")
# From here one-hot encoding starts
df_encoded = pd.get_dummies(df_label,columns=['City'])

print('\n One-Hot encoded data (city)')
print(df_encoded)
```
