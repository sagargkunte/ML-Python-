```python
from sklearn.linear_model import LinearRegression
import matplotlib.pyplot as plt

X = [[1],[2],[3],[4],[5]]
y = [40,50,60,70,80]
model = LinearRegression()

model.fit(X,y)
hours = float(input("Enter how many hours you studied : "))

predicted_marks = model.predict([[hours]])
print(f"Based on your hours of study {hours} you may score around {predicted_marks[0]} marks")

y_pred = model.predict(X)

# print(f"Y_predict is ",y_pred,X)

plt.scatter(X,y,label="Actual Data")

plt.plot(X,y_pred,label="Regression line")

plt.scatter(hours,predicted_marks[0],label="Predicted point")

plt.xlabel("Hours studied")
plt.ylabel("Marks or Score")
plt.title("Study hours vs Marks")

plt.legend()
plt.show()
```
