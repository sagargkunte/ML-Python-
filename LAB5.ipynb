```python
import pandas as pd
import numpy as np
from sklearn.neighbors import KNeighborsClassifier
from sklearn.metrics import accuracy_score
import warnings 
warnings.filterwarnings('ignore')
```

```python
np.random.seed(42)
values = np.random.rand(100)
labels = []
for i in values[:50]:
    if i <= 0.5:
        labels.append('class1')
    else:
        labels.append('class2')
labels += [None]*50

data = {
    'point':[f"{i+1}" for i in range(100)],
    'value': values,
    'label': labels
}
```

```python
df = pd.DataFrame(data)
labelled_df = df[df['label'].notna()]
x_train = labelled_df[['value']]
y_train = labelled_df[['label']]

unlabelled_df = df[df['label'].isna()]
x_test = unlabelled_df[['value']]

true_labels = []
for i in values[50]:
    if i <= 0.5:
        true_labels.append('class1')
    else:
        true_labels.append('class2')

k_values = [1,2,3,4,20,30]

for k in k_values:
    knn = KNeighborsClassifier(n_neighbors=k)
    knn.fit(x_train,y_train)
    predictions = knn.predict(x_test)
    accuracy = accuracy_score(true_labels,predictions) * 100
    print(f"Accuracy for k = {k} is {accuracy:.2f}%")
```
