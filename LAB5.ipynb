```python
import numpy as np 
import pandas as pd 
from sklearn.neighbors import KNeighborsClassifier
from sklearn.metrics import accuracy_score
np.random.seed(42)
x = np.random.rand(100)

labels = []
for i in range(50):
    if x[i] <= 0.5:
        labels.append("class1")
    else:
        labels.append("class2")
labels += [None]*50 

data = {
    "point": [f"{i+1}" for i in range(100)],
    "value": x,
    "label": labels
}

df = pd.DataFrame(data)

labelled_data = df[df["label"].notna()]
x_train = labelled_data[["value"]]
y_train = labelled_data["label"]

unlabelled_data = df[df["label"].isna()]
x_test = unlabelled_data[["value"]]

true_labels = []
for i in range(50,100): 
    if x[i] <= 0.5:
        true_labels.append("class1")
    else:
        true_labels.append("class2")

k_values = [1,2,3,4,5,20,30]

for k in k_values:
    knn = KNeighborsClassifier(n_neighbors=k)
    knn.fit(x_train,y_train)
    prediction = knn.predict(x_test)
    print(accuracy_score(true_labels,prediction)*100)
```
