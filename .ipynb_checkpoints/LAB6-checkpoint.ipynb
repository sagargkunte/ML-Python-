```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

x = np.array([5,1,2,1])
y = np.array([25,5,7,8])
x0 = 1
tau = 0.4

X = np.column_stack((np.ones(len(x)),x))
print(X)

weights = np.exp(-(x-x0)**2)/(2*tau**25)
print(weights)

W = np.diag(weights)
print(W)

X_T_W = np.dot(X.T,W)
X_T_W_X_inv = np.linalg.inv(np.dot(X_T_W,X))
beta_hat = np.dot(np.dot(X_T_W_X_inv,X_T_W),y)

y0 = np.dot([1,x0],beta_hat)
print(y0)
```
