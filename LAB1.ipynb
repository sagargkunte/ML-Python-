```python
import seaborn as sns 
import pandas as pd 
import matplotlib.pyplot as plt 
from sklearn.datasets import fetch_california_housing

housing = fetch_california_housing()
df = pd.DataFrame(housing.data,columns=housing.feature_names)
print(df)

print(df.isnull().sum())

df["HouseAge"].fillna(df["HouseAge"].median())

print("1 ---------------")
print(df.info())
print("2 ----------------")
for i in df.iloc[:]: 
    df[i] = df[i].astype('int')
print("3 -----------------")
df.info()

df.describe()

Numericals = df.select_dtypes(include="number").columns 
print(Numericals)

for col in Numericals:
    plt.figure(figsize=(10,6))
    df[col].plot(kind='hist',bins=60)
    plt.show()
print("-------------------")
plt.figure(figsize=(10,6))
sns.boxplot(data=df)

def find_outliers():
    q1 = df.quantile(0.25)
    q3 = df.quantile(0.75)
    IQR = q3-q1
    outliers = df[(df<(q1-(1.5*IQR))) | (df>(q3-(1.5*IQR)))]
    return outliers

print("Outliers are ",len(str(find_outliers())))
```
