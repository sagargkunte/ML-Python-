```python
from sklearn.linear_model import LogisticRegression

X = [[1],[2],[3],[4],[5]] # hours studied
y = [0,0,1,1,1] # result

model = LogisticRegression()
model.fit(X,y)

hours = float(input("Enter how many hours you studied : "))
result = model.predict([[hours]])[0]

if result == 1:
    print(f"Based on the hours you have studied you are likely to PASS!")
else:
    print(f"Based on the hours you have studied you are likely to FAIL!")
```
