```python
import pandas as pd 

data = {
    'Name':['sagar','kapil','lalith','ishan','ram'],
    'Age':[20,None,44,23,None],
    'Salary':[50000,60000,70000,None,None]
}
print("--------------")
df = pd.DataFrame(data)
print(df.isnull().mean()*100)
print("--------------")
print("Original DataFrame")
print(df)
print("--------------")
print(df.isnull().sum())

df_drop = df.dropna() # Drop the tables which have null values or NaN
print("--------------")
print(df_drop)

df['Age'] = df['Age'].fillna(df['Age'].mean())
df['Salary'] = df['Salary'].fillna(df['Salary'].mean())
print("--------------")
print(df)
```
