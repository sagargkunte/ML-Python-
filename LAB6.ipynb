```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

x = np.array([5,1,2,1])
y = np.array([25,5,7,8])
x0 = 1
tau = 0.4

X = np.column_stack((np.ones(len(x)),x))
print(X)
print("---------------------------------")
weights = np.exp(-(x-x0)**2)/(2*tau**25)
print(weights)
print("---------------------------------")
W = np.diag(weights)
print(W)
print("---------------------------------")
X_T_W = np.dot(X.T,W)
X_T_W_X_inv = np.linalg.inv(np.dot(X_T_W,X))
beta_hat = np.dot(np.dot(X_T_W_X_inv,X_T_W),y)

y0 = np.dot([1,x0],beta_hat)
print(y0)
print("---------------------------------")
plt.figure(figsize=(10,6))
plt.scatter(x,y,color="red",label="Data points")
plt.scatter(x0,y0,color="green",label=f"Predicted Score at x0 = {x0}")

x_range = np.linspace(min(x)-1,max(x)+1,100)
y_range = np.array([np.dot(np.array([1, xi]), beta_hat) for xi in x_range])
print(x_range)
print("---------------------------------")
print(y_range)
plt.plot(x_range,y_range,label="Locally weighted regression")


plt.legend()
plt.show()
```
